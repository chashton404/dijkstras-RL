# Dijkstra's Algorithm Game combined with Reinforcement Learning

The intent of this project was to first, make a game where a user can compare their performance on a finding the shortest path between two nodes on a map to that of Dijkstras Algorithm. After that I decided to take it one step further, I created a custom gym environment with this game, and trained a an agent using reinforcement learning to analyze their performance.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
logs = np.load("logs/evaluations.npz")

print(logs)
```

```python
timesteps = logs["timesteps"]
mean_reward = logs["results"].mean(axis=1)
std_reward = logs["results"].std(axis=1)

plt.plot(timesteps, mean_reward, label="Mean eval rewards")
plt.fill_between(timesteps, mean_reward - std_reward, mean_reward + std_reward, alpha=0.2, label="±1 std dev")
plt.axhline(0, color="gray", linestyle="--", linewidth=0.8)
plt.xlabel("Training Steps")
plt.ylabel("Episode Reward")
plt.title("Dijikstra RL Agent Learning Curve")
plt.legend()
plt.tight_layout()
plt.show
```
